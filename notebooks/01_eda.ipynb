```python
# Sanity check: these often turn out to be constant / useless for analysis
for col in ['EmployeeCount', 'StandardHours', 'Over18']:
    print(col, df[col].unique())
```

```python
df = df.drop(columns=['EmployeeCount', 'StandardHours', 'Over18', 'EmployeeNumber'])
df.shape
```

```python
attrition_rate = (df['Attrition'] == 'Yes').mean() * 100
print(f"Overall attrition rate: {attrition_rate:.1f}%")
```

```python
cat_cols = ['OverTime', 'Department', 'JobRole', 'MaritalStatus', 'BusinessTravel', 'Gender']

for col in cat_cols:
    print(f"\n--- {col} ---")
    print((pd.crosstab(df[col], df['Attrition'], normalize='index') * 100).round(1))
```

```python
fig, ax = plt.subplots(figsize=(6,4))
sns.barplot(data=df, x='OverTime', y=(df['Attrition']=='Yes').astype(int), errorbar=None, ax=ax)
ax.set_ylabel('Attrition Rate')
ax.set_title('Attrition Rate by OverTime Status')
plt.show()
```

```python
num_cols = ['MonthlyIncome', 'DistanceFromHome', 'YearsAtCompany', 'Age', 'TotalWorkingYears']

fig, axes = plt.subplots(2, 3, figsize=(16, 8))
axes = axes.flatten()
for i, col in enumerate(num_cols):
    sns.boxplot(data=df, x='Attrition', y=col, ax=axes[i])
    axes[i].set_title(col)
fig.delaxes(axes[-1])  # remove unused subplot
plt.tight_layout()
plt.show()
```

```python
ordinal_cols = ['JobSatisfaction', 'EnvironmentSatisfaction', 'WorkLifeBalance', 'JobInvolvement']

fig, axes = plt.subplots(1, 4, figsize=(18, 4))
for i, col in enumerate(ordinal_cols):
    sns.histplot(data=df, x=col, hue='Attrition', multiple='fill', discrete=True, ax=axes[i])
    axes[i].set_title(col)
    axes[i].set_ylabel('Proportion')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(14, 10))
numeric_df = df.select_dtypes(include='number')
sns.heatmap(numeric_df.corr(), cmap='coolwarm', center=0, annot=False)
plt.title('Correlation Heatmap — Numeric Features')
plt.show()
```

```python
df['Attrition_binary'] = (df['Attrition'] == 'Yes').astype(int)
df[['Attrition', 'Attrition_binary']].head()
```
